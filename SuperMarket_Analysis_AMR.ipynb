```python
ventas_ciudad = df.groupby("City")["Sales"].sum()

ventas_ciudad.plot(
    kind="bar",
    title="Ventas totales por ciudad",
    xlabel="Ciudad",
    ylabel="Ventas"
)
```

```python
ventas_producto = df.groupby("Product line")["Sales"].sum().sort_values(ascending=False)

ventas_producto.plot(
    kind="bar",
    title="Ventas por línea de producto",
    xlabel="Línea de producto",
    ylabel="Ventas"
)
```

```python
df.groupby("Customer type")["Sales"].sum()
```

```python
df.groupby("Payment")["Sales"].sum()
```

```python
ventas_cliente = df.groupby("Customer type")["Sales"].sum()

ventas_cliente.plot(
    kind="bar",
    title="Ventas según tipo de cliente",
    xlabel="Tipo de cliente",
    ylabel="Ventas"
)
```

```python
ventas_pago = df.groupby("Payment")["Sales"].sum().sort_values(ascending=False)

ventas_pago.plot(
    kind="bar",
    title="Ventas según método de pago",
    xlabel="Método de pago",
    ylabel="Ventas"
)
```

```python
df.groupby("Gender")["Sales"].sum()
```

```python
ventas_genero = df.groupby("Gender")["Sales"].sum()

ventas_genero.plot(
    kind="bar",
    title="Ventas según género",
    xlabel="Género",
    ylabel="Ventas"
)
```

```python
df.groupby("Product line")["Quantity"].sum().sort_values(ascending=False)
```

```python
cantidad_producto = df.groupby("Product line")["Quantity"].sum().sort_values(ascending=False)

cantidad_producto.plot(
    kind="bar",
    title="Cantidad de unidades vendidas por línea de producto",
    xlabel="Línea de producto",
    ylabel="Unidades vendidas"
)
```

```python
df.groupby("Product line")["Sales"].sum() / df.groupby("Product line")["Quantity"].sum()
```

```python
valor_por_unidad = (
    df.groupby("Product line")["Sales"].sum()
    / df.groupby("Product line")["Quantity"].sum()
).sort_values(ascending=False)

valor_por_unidad.plot(
    kind="bar",
    title="Valor promedio de ventas por unidad",
    xlabel="Línea de producto",
    ylabel="Valor promedio por unidad"
)
```

```python
ventas_fecha = df.groupby("Date")["Sales"].sum()

ventas_fecha
```

```python
ventas_fecha.idxmax(), ventas_fecha.max()
```

```python
ventas_mes = df.groupby(df["Date"].dt.to_period("M"))["Sales"].sum()

ventas_mes
```

```python
ventas_mes.plot(
    kind="bar",
    title="Ventas totales por mes",
    xlabel="Mes",
    ylabel="Ventas"
)
```

# 📌 CONCLUSIONES

A partir del análisis realizado sobre el conjunto de datos de ventas del
supermercado, se identificaron diferentes características relacionadas con
el comportamiento de las ventas, los productos y los clientes.

En el período analizado se registraron ventas totales de **322.966,749**,
con un promedio de **322,967 por transacción** y un total de **5.510
unidades vendidas**.

La línea de productos con mayores ventas fue **Food and beverages**, con
**56.144,844**, mientras que **Health and beauty** presentó el menor valor
de ventas, con **49.193,739**.

En cuanto al ingreso bruto, se obtuvo un total de **15.379,369**. La línea
de productos con mayor ingreso bruto fue nuevamente **Food and beverages**,
con **2.673,564**.

Respecto a las calificaciones de los clientes, el promedio general fue de
**6,9727**, con una calificación máxima de **10** y una mínima de **4**.
La línea de productos con mayor calificación promedio fue **Food and
beverages**, con **7,1132**.

El análisis temporal permitió observar variaciones en las ventas durante
los meses de enero, febrero y marzo de 2019, permitiendo identificar los
períodos con mayores y menores niveles de ventas.

Finalmente, el uso de **Python, Pandas y Jupyter Notebook** permitió
organizar, procesar y analizar los datos de manera estructurada, facilitando
la identificación de patrones y la interpretación de los resultados.
